```python
bulan = input("Bulan apa anda ingin menginap (Mei, Juni, Juli, Agustus, September, Oktober): ")
lama = int(input("Berapa hari anda ingin menginap: "))
tipe = input("Tipe kamar apa yang anda inginkan? (Studio / Apartement): ")

tarif_studio = 0
tarif_apartement = 0

if bulan == "Mei" or bulan == "Oktober":
    tarif_studio = 500000
    tarif_apartement = 650000
elif bulan == "Juni" or bulan == "September":
    tarif_studio = 752000
    tarif_apartement = 687000
elif bulan == "Juli" or bulan == "Agustus":
    tarif_studio = 760000
    tarif_apartement = 770000
else:
    print("Bulan yang anda inginkan tidak tersedia")

total_tarif = 0
if tipe == "Studio":
    total_tarif = tarif_studio * lama
    if lama > 3:
        diskon = total_tarif * 0.15
        total_tarif -= diskon

elif tipe == "Apartement":
    total_tarif = tarif_apartement * lama
    if lama > 3:
        diskon = total_tarif * 0.10
        total_tarif -= diskon

else:
    print("Tipe kamar yang anda masukkan tidak valid")

if total_tarif > 0:
    print(f"\nTotal tarif untuk menginap selama {lama} hari di bulan {bulan} dengan tipe kamar {tipe} adalah Rp. {total_tarif:,.0f}")
```

```python
data_pengiriman = (
    ("Laptop", 2.5, "Surabaya", 35000),
    ("Printer", 8.0, "Jakarta", 55000),
    ("Monitor", 4.2, "Bandung", 45000),
    ("Keyboard", 1.2, "Surabaya", 30000),
    ("Mouse", 0.5, "Jakarta", 25000)
)

print("a. Data pengiriman ke Surabaya")
for barang in data_pengiriman:
    if barang[2] == "Surabaya":
        print(barang)
        
total_berat = 0
for barang in data_pengiriman:
    if barang[2] == "Jakarta":
        total_berat = total_berat + barang[1]

print("\n b. Total berat ke Jakarta")
print("Total berat:", total_berat, "kg")

hasil_ubah = []
for barang in data_pengiriman:
    nama = barang[0]
    berat = barang[1]
    kota = barang[2]
    biaya = barang[3]
    
    if kota == "Bandung":
        biaya = biaya + (biaya * 0.10)
        
    hasil_ubah.append((nama, berat, kota, biaya))

data_pengiriman_baru = tuple(hasil_ubah)

print("\n c. Data baru setelah biaya Bandung naik 10%")
print(data_pengiriman_baru)
```

```python
nama_barang = input("Nama barang: ")
jumlah_barang = input("Jumlah awal (misal: 3 lusin): ")
satuan_barang = input("Satuan tujuan (pcs / lusin / kodi): ")

bagian = jumlah_barang.split()
jumlah_awal = float(bagian[0])
satuan_awal = bagian[1].lower()

if satuan_awal == "pcs":
    total_pcs = jumlah_awal
elif satuan_awal == "lusin":
    total_pcs = jumlah_awal * 12
elif satuan_awal == "kodi":
    total_pcs = jumlah_awal * 20
else:
    total_pcs = 0

if total_pcs == 0:
    print("Error: Satuan awal tidak dikenal!")
else:
    satuan_barang = satuan_barang.lower()
    
    if satuan_barang == "pcs":
        hasil_konversi = total_pcs
    elif satuan_barang == "lusin":
        hasil_konversi = total_pcs / 12
    elif satuan_barang == "kodi":
        hasil_konversi = total_pcs / 20
    else:
        hasil_konversi = -1

    if hasil_konversi == -1:
        print("Error: Satuan barang tidak dikenal!")
    else:
        print("\nBarang          :", nama_barang)
        print("Jumlah awal     :", jumlah_barang)
        print("Total dalam pcs :", int(total_pcs), "pcs")
        print("Konversi ke", satuan_barang, ":", hasil_konversi, satuan_barang)
```
